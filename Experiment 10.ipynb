```python
import os
os.environ["OMP_NUM_THREADS"] = "1"
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans

%matplotlib inline
```

```python
df = pd.read_excel("Mall_Customers - Mall_Customers.xlsx")

df.info()
```

```python
df.head()
```

```python
sns.pairplot(df)
plt.show()
```

```python
features = df.iloc[:, [3, 4]].values

features
```

```python
model = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

model.fit(features)
```

```python
labels = model.predict(features)

labels
```

```python
Final = df.iloc[:, [3, 4]].copy()

Final["label"] = labels

Final.head()
```

```python
sns.set_style("whitegrid")

sns.FacetGrid(
    Final,
    hue="label",
    height=8
).map(
    plt.scatter,
    "Annual Income (k$)",
    "Spending Score (1-100)"
).add_legend()

plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Customer Segmentation using K-Means")

plt.show()
```

```python
features_el = df.iloc[:, [2, 3, 4]].values

features_el
```

```python
wcss = []

for i in range(1, 10):
    model = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )
    
    model.fit(features_el)
    
    wcss.append(model.inertia_)

wcss
```
